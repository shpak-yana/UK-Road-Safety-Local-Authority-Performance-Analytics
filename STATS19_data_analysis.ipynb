3. Data cleaning: replace invalid values with NaN.

```python
CODEBOOK_PATH = 'dft-road-casualty-statistics-road-safety-open-dataset-data-guide-2025.xlsx'
codebook = pd.read_excel(CODEBOOK_PATH, sheet_name='2024_code_list')
```

```python
codebook = codebook[codebook['code/format'].notna()].copy()
codebook['code_num'] = pd.to_numeric(codebook['code/format'], errors='coerce')
codebook = codebook[codebook['code_num'].notna()].copy()
```

```python
BAD_LABEL_WORDS = [
    'unknown', 'missing', 'out of range',
    'not known', 'data missing', 'not known or not requested'
]

mask_bad = codebook['label'].astype(str).str.lower().str.contains(
    '|'.join(BAD_LABEL_WORDS), na=False
)

codebook_clean = codebook[~mask_bad]
```

valid_codes contains only "normal" codes. Anything like -1, 99, 9, or 3—or anything marked as unknown/missing—will not be included.

```python
valid_codes = (
    codebook_clean
    .groupby('field name')['code_num']
    .apply(lambda s: set(s.astype(int)))
    .to_dict()
)
```

```python
NUMERIC_SENTINELS = {
    'age_of_casualty': [-1],
    'age_of_driver': [-1],
    'engine_capacity_cc': [-1],
    'age_of_vehicle': [-1],
    'speed_limit': [-1, 99],       # 99 = unknown (self reported)
    'first_road_number': [-1],
    'second_road_number': [-1],
}
```

```python
def clean_stats19(df, valid_codes, numeric_sentinels):
    df_clean = df.copy()
    
    # 1. Categorical fields: keep only valid codes
    for col, valid in valid_codes.items():
        if col in df_clean.columns:
            df_clean.loc[~df_clean[col].isin(valid), col] = np.nan
    
    # 2. Numeric fields: replace -1 / 99 / other sentinel values with NaN
    for col, sentinels in numeric_sentinels.items():
        if col in df_clean.columns:
            df_clean.loc[df_clean[col].isin(sentinels), col] = np.nan
            
    return df_clean
```

```python
casualty_clean = clean_stats19(casualty_df, valid_codes, NUMERIC_SENTINELS)
collision_clean = clean_stats19(collision_df, valid_codes, NUMERIC_SENTINELS)
vehicle_clean = clean_stats19(vehicle_df, valid_codes, NUMERIC_SENTINELS)
```

```python
collision_clean = collision_clean[collision_clean['police_force'].between(1, 63)].copy()

# Парсинг даты
collision_clean['date'] = pd.to_datetime(collision_clean['date'],dayfirst=True, errors='coerce')
collision_clean['year'] = collision_clean['date'].dt.year
collision_clean['month'] = collision_clean['date'].dt.month
collision_clean['hour'] = pd.to_numeric(collision_clean['time'].str[:2], errors='coerce')
```

```python
# Сокращённый набор столбцов
cols = [
        'collision_index', 'collision_year', 'date', 'year', 'month', 'hour',
        'police_force', 'local_authority_district', 'local_authority_ons_district',
        'longitude', 'latitude', 'road_type', 'urban_or_rural_area',
        'light_conditions', 'weather_conditions', 'road_surface_conditions',
        'first_road_class', 'speed_limit', 'junction_detail',
        'number_of_vehicles', 'number_of_casualties',
        'collision_severity',
    ]
collision_clean = collision_clean[[c for c in cols if c in collision_clean.columns]]

collision_clean.to_parquet('data/processed/collisions.parquet', index=False)
```

```python
lsoa_col = 'lsoa_of_casualty' if 'lsoa_of_casualty' in casualty_clean.columns \
               else 'enhanced_casualty_lsoa'

casualty_clean = casualty_clean.rename(columns={lsoa_col: 'lsoa_of_casualty'})

# Убираем строки без LSOA
casualty_clean = casualty_clean.dropna(subset=['lsoa_of_casualty'])
casualty_clean = casualty_clean[casualty_clean['lsoa_of_casualty'].str.startswith('E', na=False)]

# Приведение возраста к числу
if 'age_of_casualty' in casualty_clean.columns:
    casualty_clean['age_of_casualty'] = pd.to_numeric(casualty_clean['age_of_casualty'], errors='coerce')

cols = [
        'collision_index', 'vehicle_reference', 'casualty_reference',
        'casualty_class', 'sex_of_casualty', 'age_of_casualty',
        'age_band_of_casualty', 'casualty_severity', 'pedestrian_location',
        'pedestrian_movement', 'car_passenger', 'bus_or_coach_passenger',
        'pedestrian_road_maintenance_worker', 'casualty_type',
        'casualty_home_area_type', 'casualty_imd_decile',
        'lsoa_of_casualty',
    ]

casualty_clean = casualty_clean[[c for c in cols if c in casualty_clean.columns]]

casualty_clean.to_parquet('data/processed/casualties.parquet', index=False)
```

```python
# Определяем имя поля LSOA водителя (в разных версиях разное)
driver_lsoa_col = next(
        (c for c in ['lsoa_of_driver', 'enhanced_driver_lsoa', 'driver_lsoa']
         if c in vehicle_clean.columns),
        None
    )

if driver_lsoa_col:
    vehicle_clean = vehicle_clean.rename(columns={driver_lsoa_col: 'lsoa_of_driver'})

 # Приведение числовых полей
for col in ['age_of_driver', 'age_of_vehicle', 'engine_capacity_cc',
            'driver_imd_decile', 'sex_of_driver', 'vehicle_type']:
    if col in vehicle_clean.columns:
        vehicle_clean[col] = pd.to_numeric(vehicle_clean[col], errors='coerce')

cols = [
        'collision_index', 'vehicle_reference',
        'vehicle_type', 'vehicle_manoeuvre',
        'vehicle_location_restricted_lane', 'junction_location',
        'skidding_and_overturning', 'hit_object_in_carriageway',
        'vehicle_leaving_carriageway', 'hit_object_off_carriageway',
        'first_point_of_impact', 'vehicle_left_hand_drive',
        'journey_purpose_of_driver', 'sex_of_driver',
        'age_of_driver', 'age_band_of_driver',
        'engine_capacity_cc', 'propulsion_code', 'age_of_vehicle',
        'driver_imd_decile', 'driver_home_area_type',
        'driver_distance_banding',
        'lsoa_of_driver',
    ]

vehicle_clean = vehicle_clean[[c for c in cols if c in vehicle_clean.columns]]

vehicle_clean.to_parquet('data/processed/vehicles.parquet', index=False)
```
